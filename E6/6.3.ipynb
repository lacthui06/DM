```python
# 1. Nạp thư viện FIM và dữ liệu giao dịch Market-Basket
from fim import eclat, fpgrowth

dataset = [
    ['Bread', 'Milk'],
    ['Bread', 'Diaper', 'Beer', 'Eggs'],
    ['Milk', 'Diaper', 'Beer', 'Coke'],
    ['Bread', 'Milk', 'Diaper', 'Beer'],
    ['Bread', 'Milk', 'Diaper', 'Coke']
]

N = len(dataset)
min_sup = 40   # 40% (tối thiểu 2/5 giao dịch)
min_conf = 60  # 60%

print(f"Số lượng giao dịch: {N}")
print(f"Cấu hình tham số: minSup = {min_sup}% (tối thiểu {int(min_sup*N/100)}/{N} giao dịch), minConf = {min_conf}%")
```

```python
# 2. Khai phá các tập phổ biến (Frequent Itemsets) bằng giải thuật ECLAT
# - ECLAT sử dụng định dạng dữ liệu dọc (Vertical Data Format - TID-list)
# - supp=40: độ hỗ trợ tối thiểu 40%
# - zmin=1: kích thước tập mục tối thiểu từ 1 phần tử
# - report='as': xuất số lượng giao dịch tuyệt đối (a) và tỷ lệ hỗ trợ (s)
frequent_eclat = eclat(dataset, supp=min_sup, zmin=1, report='as')

# Phân loại và in theo từng cấp độ k
by_k_eclat = {}
for itemset, sup_cnt, sup_frac in frequent_eclat:
    k = len(itemset)
    by_k_eclat.setdefault(k, []).append((itemset, sup_cnt, sup_frac))

print(f"=== KẾT QUẢ TẬP PHỔ BIẾN BẰNG ECLAT (Tổng: {len(frequent_eclat)} tập) ===")
for k in sorted(by_k_eclat.keys()):
    print(f"\nFrequent {k}-itemset ({len(by_k_eclat[k])} tập):")
    for itemset, sup_cnt, sup_frac in sorted(by_k_eclat[k], key=lambda x: x[1], reverse=True):
        print(f"  {set(itemset)} | Support = {sup_frac:.2f} ({sup_cnt}/{N}) | {sup_frac*100:.0f}%")
print("=" * 65)
```

```python
# 3. Khai phá các tập phổ biến (Frequent Itemsets) bằng giải thuật FP-GROWTH
# - FP-Growth nén cơ sở dữ liệu vào cây FP-Tree và khai phá tập phổ biến không cần sinh ứng viên
frequent_fp = fpgrowth(dataset, supp=min_sup, zmin=1, report='as')

# Phân loại và in theo từng cấp độ k
by_k_fp = {}
for itemset, sup_cnt, sup_frac in frequent_fp:
    k = len(itemset)
    by_k_fp.setdefault(k, []).append((itemset, sup_cnt, sup_frac))

print(f"=== KẾT QUẢ TẬP PHỔ BIẾN BẰNG FP-GROWTH (Tổng: {len(frequent_fp)} tập) ===")
for k in sorted(by_k_fp.keys()):
    print(f"\nFrequent {k}-itemset ({len(by_k_fp[k])} tập):")
    for itemset, sup_cnt, sup_frac in sorted(by_k_fp[k], key=lambda x: x[1], reverse=True):
        print(f"  {set(itemset)} | Support = {sup_frac:.2f} ({sup_cnt}/{N}) | {sup_frac*100:.0f}%")
print("=" * 65)
```

```python
# 4. Sinh các luật kết hợp (Association Rules) bằng Eclat và FP-Growth (target='r')
# - supp=40: support tối thiểu 40%
# - conf=60: confidence tối thiểu 60%
# - zmin=2: tối thiểu 2 mục (tiền đề có ít nhất 1 mục)
# - report='asCl': xuất absolute support (a), support fraction (s), confidence % (C), lift (l)
rules_eclat = eclat(dataset, target='r', supp=min_sup, conf=min_conf, zmin=2, report='asCl')
rules_fp = fpgrowth(dataset, target='r', supp=min_sup, conf=min_conf, zmin=2, report='asCl')

# Sắp xếp các luật theo Confidence giảm dần, sau đó theo Lift và Support
sorted_rules_fp = sorted(rules_fp, key=lambda x: (x[4], x[5], x[3]), reverse=True)

print(f"DANH SÁCH CÁC LUẬT KẾT HỢP (minConf = {min_conf}%, minSup = {min_sup}%) - Tổng: {len(sorted_rules_fp)} luật:\n")
for idx, (head, body, sup_cnt, sup_frac, conf_pct, lift_val) in enumerate(sorted_rules_fp, 1):
    ant_str = str(set(body))
    con_str = str(set([head]))
    print(f"{idx:2d}. {ant_str:<25} => {con_str:<12} | Support = {sup_frac:.2f} ({sup_cnt}/{N}) | Conf = {conf_pct:5.1f}% | Lift = {lift_val:.3f}")
```

```python
# 5. So sánh đối chiếu kết quả giữa Eclat, FP-Growth và Apriori
set_eclat = {tuple(sorted(x[0])): (x[1], round(x[2], 2)) for x in frequent_eclat}
set_fp = {tuple(sorted(x[0])): (x[1], round(x[2], 2)) for x in frequent_fp}

print("=== ĐỐI CHIẾU KẾT QUẢ GIỮA CÁC GIẢI THUẬT ===")
print(f"-> Số tập phổ biến tìm được bởi ECLAT     : {len(frequent_eclat)}")
print(f"-> Số tập phổ biến tìm được bởi FP-GROWTH : {len(frequent_fp)}")
print(f"-> Kết quả hai giải thuật trùng khớp 100% : {set_eclat == set_fp}")
print(f"-> Số luật kết hợp tìm được bởi ECLAT    : {len(rules_eclat)}")
print(f"-> Số luật kết hợp tìm được bởi FP-GROWTH: {len(rules_fp)}")
print("Cả Eclat và FP-Growth đều cho kết quả hoàn toàn đồng nhất với Apriori (Bài 6.1 & 6.2)!")
```
