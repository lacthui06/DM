```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans

# 1. Khởi tạo tập dữ liệu 2 chiều (đồng bộ với bài 4.1)
df = pd.DataFrame({
    'x': [14, 22, 19, 27, 30, 35, 24, 46, 48, 53, 55, 58, 51, 56, 59, 64, 66, 71, 75, 68],
    'y': [38, 35, 50, 33, 53, 45, 57, 61, 69, 73, 65, 60, 67, 21, 15, 9, 22, 8, 26, 17]
})

# 2. Huấn luyện mô hình KMeans từ Scikit-Learn
kmeans = KMeans(n_clusters=3, n_init=10, random_state=42)
labels = kmeans.fit_predict(df)
centroids = kmeans.cluster_centers_

# 3. Định nghĩa bảng màu cho từng cụm
colmap = {0: 'red', 1: 'green', 2: 'blue'}
colors = [colmap[lbl] for lbl in labels]

# 4. Trực quan hóa kết quả phân cụm và các tâm cụm
plt.figure(figsize=(5, 5))
plt.scatter(df['x'], df['y'], color=colors, alpha=0.6, edgecolor='k', s=45)

# Vẽ các tâm cụm bằng dấu sao/kim cương lớn
for idx, centroid in enumerate(centroids):
    plt.scatter(centroid[0], centroid[1], color=colmap[idx], marker='D', s=130, edgecolor='black')

plt.xlim(5, 85)
plt.ylim(0, 85)
plt.title('Kết quả phân cụm K-Means (Scikit-Learn)', fontsize=11, fontweight='bold')
plt.xlabel('Tọa độ X')
plt.ylabel('Tọa độ Y')
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```
