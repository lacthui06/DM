## 3. Các kiểu dữ liệu nâng cao: List, Tuple, Set, Dictionary

### 3.1 Danh sách (`List`)
- **Mô tả ngắn:** Tập hợp các phần tử có thứ tự, có thể thay đổi giá trị (mutable), cho phép các phần tử trùng lặp. Khai báo bằng ngoặc vuông `[...]`.
- **Ngữ cảnh áp dụng:** Lưu trữ danh sách phần tử biến động cần thêm, sửa, xóa linh hoạt (danh sách sinh viên, mảng dữ liệu thu thập).

### 3.2 Bộ dữ liệu cố định (`Tuple`)
- **Mô tả ngắn:** Tập hợp các phần tử có thứ tự, KHÔNG thể thay đổi giá trị (immutable) sau khi tạo. Khai báo bằng ngoặc đơn `(...)`.
- **Ngữ cảnh áp dụng:** Lưu trữ dữ liệu cố định tránh bị thay đổi vô ý trong quá trình chạy (tọa độ GPS, kích thước ảnh, hằng số cấu hình).

### 3.3 Tập hợp (`Set`)
- **Mô tả ngắn:** Tập hợp không có thứ tự, KHÔNG chứa phần tử trùng lặp (các giá trị là duy nhất). Khai báo bằng ngoặc nhọn `{...}`.
- **Ngữ cảnh áp dụng:** Loại bỏ phần tử trùng lặp trong danh sách, thực hiện các phép toán tập hợp (giao, hợp, hiệu).

### 3.4 Từ điển (`Dictionary`)
- **Mô tả ngắn:** Lưu trữ dữ liệu dạng cặp Khóa - Giá trị (`key: value`). Khóa là duy nhất và bất biến. Khai báo bằng ngoặc `{key: value}`.
- **Ngữ cảnh áp dụng:** Tra cứu dữ liệu nhanh theo khóa, biểu diễn đối tượng có nhiều thuộc tính (hồ sơ sinh viên, định dạng dữ liệu JSON).


```python
# Áp dụng các phương thức và thao tác cho từng kiểu dữ liệu nâng cao

# 1. LIST (Mutable - Thêm, sửa, xóa, sắp xếp)
diem_so = [8.5, 7.0, 9.2, 6.8]
diem_so.append(9.5)             # Thêm vào cuối
diem_so.insert(1, 8.0)          # Chèn vào vị trí chỉ số 1
diem_so[0] = 8.8                # Sửa giá trị theo chỉ số index
diem_so.sort(reverse=True)      # Sắp xếp giảm dần tại chỗ
print("=== 1. DANH SÁCH (List) ===")
print(f"List điểm sau khi biến đổi: {diem_so}")
print(f"Phần tử lớn nhất (max): {max(diem_so)} | Phần tử nhỏ nhất (min): {min(diem_so)}\n")

# 2. TUPLE (Immutable - Đọc, đếm, giải nén dữ liệu cố định)
toa_do_uth = (10.7769, 106.7009, "Cơ sở chính UTH")
lat, long, ten_cs = toa_do_uth  # Tuple Unpacking (Giải nén tuple)
tuple_sample = (1, 2, 2, 3, 2, 4)
print("=== 2. BỘ DỮ LIỆU CỐ ĐỊNH (Tuple) ===")
print(f"Tuple ban đầu: {toa_do_uth}")
print(f"Giải nén Tuple => Vĩ độ: {lat}, Kinh độ: {long}, Địa điểm: '{ten_cs}'")
print(f"Số lần xuất hiện số 2 trong tuple_sample (.count()): {tuple_sample.count(2)}")
print(f"Vị trí xuất hiện đầu tiên của số 3 (.index()): {tuple_sample.index(3)}\n")

# 3. SET (Unordered, Unique - Phép toán tập hợp & Lọc trùng)
ds_trung = ["Python", "Java", "Python", "SQL", "Java", "C++"]
set_doc_nhat = set(ds_trung)     # Lọc trùng lập tức
set_a = {"Python", "R", "SQL", "Machine Learning"}
set_b = {"Python", "Java", "C++", "SQL"}
print("=== 3. TẬP HỢP (Set) ===")
print(f"Danh sách ban đầu (có trùng): {ds_trung}")
print(f"Set sau khi lọc phần tử trùng: {set_doc_nhat}")
print(f"Giao giữa set_a & set_b (.intersection()): {set_a.intersection(set_b)}")
print(f"Hợp giữa set_a | set_b (.union()): {set_a.union(set_b)}")
print(f"Hiệu set_a - set_b (.difference()): {set_a.difference(set_b)}\n")

# 4. DICTIONARY (Key-Value - Tra cứu siêu tốc, phương thức .get() an toàn)
sinh_vien = {
    "ma_sv": "089206000860",
    "ho_ten": "Thái Anh Lạc",
    "lop": "012012204103",
    "diem_tb": 8.6
}
sinh_vien["xep_loai"] = "Giỏi"   # Thêm key mới
sinh_vien["diem_tb"] = 8.8      # Cập nhật giá trị key cũ

print("=== 4. TỪ ĐIỂN (Dictionary) ===")
print(f"Thông tin sinh viên: {sinh_vien}")
print(f"Truy cập an toàn bằng .get('email'): {sinh_vien.get('email', 'Chưa đăng ký email')}")
print(f"Danh sách các Key (.keys()): {list(sinh_vien.keys())}")
print(f"Danh sách các Value (.values()): {list(sinh_vien.values())}")
print("Duyệt từng cặp Key - Value (.items()):")
for k, v in sinh_vien.items():
    print(f"  + {k}: {v}")
```

## 4. Cấu trúc lệnh trong Python

### 4.a Lệnh điều kiện


```python
# Ví dụ 4.a: Lệnh điều kiện if - elif - else
diem_tb = 8.6

if diem_tb >= 9.0:
    xep_loai = "Xuất sắc"
elif diem_tb >= 8.0:
    xep_loai = "Giỏi"
elif diem_tb >= 6.5:
    xep_loai = "Khá"
elif diem_tb >= 5.0:
    xep_loai = "Trung bình"
else:
    xep_loai = "Yếu/Kém"

print(f"Điểm TB: {diem_tb} => Xếp loại học lực: {xep_loai}")
```

### 4.b Lệnh lặp


```python
# Ví dụ 4.b: Lệnh lặp for và while

# Duyệt danh sách dùng vòng lặp for
danh_sach_diem = [7.5, 8.0, 9.2, 6.5, 8.8]
tong_diem = 0
for diem in danh_sach_diem:
    tong_diem += diem

print(f"Danh sách điểm: {danh_sach_diem}")
print(f"Điểm trung bình (dùng for): {tong_diem / len(danh_sach_diem):.2f}\n")

# Lặp dùng while
n, tong_n, i = 5, 0, 1
while i <= n:
    tong_n += i
    i += 1
print(f"Tổng từ 1 đến {n} (dùng while): {tong_n}")
```

## 5. Khai báo hàm


```python
# Ví dụ 5: Khai báo hàm dùng def
def tinh_thong_ke_diem(danh_sach_diem, lam_tron=2):
    """Hàm tính toán điểm trung bình, max, min từ danh sách điểm đầu vào."""
    if not danh_sach_diem:
        return None
    
    return {
        "trung_binh": round(sum(danh_sach_diem) / len(danh_sach_diem), lam_tron),
        "cao_nhat": max(danh_sach_diem),
        "thap_nhat": min(danh_sach_diem),
        "so_luong": len(danh_sach_diem)
    }

# Gọi hàm
bang_diem = [8.5, 9.0, 7.5, 6.0, 9.5, 8.0]
ket_qua = tinh_thong_ke_diem(bang_diem)
print("=== THỐNG KÊ ĐIỂM ===")
print(f"Danh sách: {bang_diem}")
print(f"Kết quả thống kê: {ket_qua}")
```
